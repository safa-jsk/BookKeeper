```python
import pandas as pd
import numpy as np
```

```python
books = pd.read_csv('books.csv')
```

```python
books.info()
```

```python
books.nunique()
```

```python
books.isnull().sum()
```

```python
books = books.dropna()
```

```python
books.isnull().sum()
```

```python
books.nunique()
```

```python
books['summary'] = books['summary'].apply(lambda x:x.split())
```

```python
books.head()
```

```python
books['author'] = books['author'].apply(lambda x:x.replace(" ", ""))
```

```python
books.head()
```

```python
books['genre'] = books['genre'].apply(lambda x:x.split(","))
```

```python
books['author'] = books['author'].apply(lambda x:[x])
```

```python
books.head(40)
```

```python
books['tags'] = books['author'] + books['genre'] + books['summary']
```

```python
new_df = books[['title', 'tags']]
```

```python
new_df.head()
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:" ".join(x))
```

```python
new_df.head()
```

```python
new_df['tags'].apply(lambda x:x.lower())
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))

    s = " ".join(y)
    return s
```

```python
new_df['tags'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
vectors.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
def recommend(book):
    book_index = new_df[new_df['title'] == book].index[0]
    distances = similarity[book_index]
    book_list = sorted(list(enumerate(distances)), reverse = True, key = lambda x:x[1])[1:6]

    for i in book_list:
        print(new_df.iloc[i[0]].title)
```

```python
import pickle
```

```python
pickle.dump(new_df.to_dict(), open('booksAI.pkl', 'wb'))
```

```python
pickle.dump(similarity, open('similarity.pkl', 'wb'))
```

```python
recommend('Harry Potter and the Chamber of Secrets')
```
